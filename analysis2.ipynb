```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv('/content/flood_risk_dataset_india_cleaned.1 (1).csv')

print("DATASET SHAPE:", df.shape)

# Q1. What percentage of records experienced floods?
print("\nQ1. Flood and non-flood percentages")
print(df['flood_occurred'].value_counts())
print(df['flood_occurred'].value_counts(normalize=True).mul(100).round(2))

# Q2. Is rainfall higher when floods occur?
print("\nQ2. Average rainfall by flood occurrence")
print(df.groupby('flood_occurred')['rainfall__mm_'].mean().round(2))

# Q3. Is river discharge higher during floods?
print("\nQ3. Average river discharge")
print(df.groupby('flood_occurred')['river_discharge__m³_s_'].mean().round(2))

# Q4. How does water level differ during floods?
print("\nQ4. Average water level")
print(df.groupby('flood_occurred')['water_level__m_'].mean().round(2))

# Q5. Which soil type has the highest flood percentage?
print("\nQ5. Flood percentage by soil type")
print(df.groupby('soil_type')['flood_occurred'].mean().mul(100).sort_values(ascending=False).round(2))

# Q6. Which land-cover type has the highest flood percentage?
print("\nQ6. Flood percentage by land cover")
print(df.groupby('land_cover')['flood_occurred'].mean().mul(100).sort_values(ascending=False).round(2))

# Q7. Is historical flood information related to current floods?
print("\nQ7. Historical floods vs current flood occurrence (%)")
print(pd.crosstab(df['historical_floods'], df['flood_occurred'], normalize='index').mul(100).round(2))

# Q8. How does population density differ by flood occurrence?
print("\nQ8. Population density comparison")
print(df.groupby('flood_occurred')['population_density'].agg(['mean', 'median']).round(2))

# Q9. Which elevation range has the highest flood percentage?
print("\nQ9. Flood percentage by elevation range")
elevation_group = pd.qcut(df['elevation__m_'], q=3, duplicates='drop')
print(df.groupby(elevation_group, observed=True)['flood_occurred'].mean().mul(100).round(2))

# Q10. Which numerical factors correlate with flood occurrence?
print("\nQ10. Correlation with flood occurrence")
cols = ['flood_occurred', 'rainfall__mm_', 'water_level__m_',
        'river_discharge__m³_s_', 'temperature__c_', 'humidity__',
        'elevation__m_', 'population_density']
print(df[cols].corr()['flood_occurred'].sort_values(ascending=False).round(3))

# Q11. How does flood percentage change with rainfall ranges?
print("\nQ11. Flood percentage by rainfall range")
rain_group = pd.qcut(df['rainfall__mm_'], q=3, duplicates='drop')
print(pd.crosstab(rain_group, df['flood_occurred'], normalize='index').mul(100).round(2))

# Q12. How does temperature relate to floods?
print("\nQ12. Flood percentage by temperature range")
temp_group = pd.qcut(df['temperature__c_'], q=3, duplicates='drop')
print(pd.crosstab(temp_group, df['flood_occurred'], normalize='index').mul(100).round(2))

# Q13. How does humidity relate to floods?
print("\nQ13. Flood percentage by humidity range")
humidity_group = pd.qcut(df['humidity__'], q=3, duplicates='drop')
print(pd.crosstab(humidity_group, df['flood_occurred'], normalize='index').mul(100).round(2))

# Q14. Is rainfall related to water level?
print("\nQ14. Rainfall and water-level correlation")
print(df[['rainfall__mm_', 'water_level__m_']].corr().round(3))

# Q15. Which soil and land-cover combinations have the highest flood percentage?
print("\nQ15. Top 10 soil and land-cover combinations by flood percentage")
result = pd.crosstab([df['soil_type'], df['land_cover']],
                     df['flood_occurred'], normalize='index').mul(100)
if 1 in result.columns:
    print(result.sort_values(by=1, ascending=False).head(10).round(2))
else:
    print("No flood category coded as 1 was found. Check flood_occurred values.")

print("\nALL 15 ANALYSIS QUESTIONS COMPLETED!")
```
