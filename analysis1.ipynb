```python
import pandas as pd
import numpy as np


df = pd.read_csv('flood_risk_dataset_india_cleaned (1).csv')


print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())


print("\nStatistical Summary:")
print(df.describe())


print("\nFlood Occurrence:")
print(df['flood_occurred'].value_counts())

print("\nFlood Occurrence Percentage:")
print(df['flood_occurred'].value_counts(normalize=True) * 100)

print("\nAverage Rainfall by Flood Occurrence:")
print(df.groupby('flood_occurred')['rainfall__mm_'].mean())


print("\nAverage Water Level by Flood Occurrence:")
print(df.groupby('flood_occurred')['water_level__m_'].mean())


print("\nAverage River Discharge by Flood Occurrence:")
print(df.groupby('flood_occurred')['river_discharge__m³_s_'].mean())


print("\nAverage Temperature by Flood Occurrence:")
print(df.groupby('flood_occurred')['temperature__c_'].mean())


print("\nAverage Humidity by Flood Occurrence:")
print(df.groupby('flood_occurred')['humidity__'].mean())


print("\nAverage Elevation by Flood Occurrence:")
print(df.groupby('flood_occurred')['elevation__m_'].mean())


print("\nAverage Population Density by Flood Occurrence:")
print(df.groupby('flood_occurred')['population_density'].mean())


print("\nHistorical Floods:")
print(df['historical_floods'].value_counts())

print("\nHistorical Floods vs Flood Occurrence:")
print(pd.crosstab(df['historical_floods'], df['flood_occurred']))


print("\nLand Cover Distribution:")
print(df['land_cover'].value_counts())

print("\nLand Cover vs Flood Occurrence:")
print(pd.crosstab(df['land_cover'], df['flood_occurred']))


print("\nSoil Type Distribution:")
print(df['soil_type'].value_counts())

print("\nSoil Type vs Flood Occurrence:")
print(pd.crosstab(df['soil_type'], df['flood_occurred']))


print("\nInfrastructure vs Flood Occurrence:")
print(df.groupby('flood_occurred')['infrastructure'].mean())


rainfall = df['rainfall__mm_'].to_numpy()

print("\nNumPy Rainfall Analysis:")
print("Minimum:", np.min(rainfall))
print("Maximum:", np.max(rainfall))
print("Mean:", np.mean(rainfall))
print("Median:", np.median(rainfall))
print("Standard Deviation:", np.std(rainfall))


numeric_df = df.select_dtypes(include=np.number)

correlation = numeric_df.corr()

print("\nCorrelation Matrix:")
print(correlation)


print("\nCorrelation with Flood Occurred:")
print(correlation['flood_occurred'].sort_values(ascending=False))


print("\nFinal Flood Analysis Summary:")

summary = df.groupby('flood_occurred')[
    [
        'rainfall__mm_',
        'temperature__c_',
        'humidity__',
        'river_discharge__m³_s_',
        'water_level__m_',
        'elevation__m_',
        'population_density'
    ]
].mean()

print(summary)

print("\nDATA ANALYSIS COMPLETED!")
```
